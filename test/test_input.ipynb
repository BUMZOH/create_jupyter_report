# input() による日付入力テスト

## 概要

Python標準の `input()` を使用して、開始日と終了日を入力するサンプル。

Notebookのソースコードを変更せずに、実行時に日付を指定できる。

## 動作仕様

- VS Codeの画面上部に入力ボックスが表示される。
- 開始日と終了日を順番に入力する。
- 入力形式は `YYYY-MM-DD` とする。
- 何も入力せずEnterを押すと、今日の日付が採用される。
- 入力値は文字列として `start_date`、`end_date` に格納される。

## コードのポイント

- `date.today()`：今日の日付を取得する。
- `isoformat()`：日付を `YYYY-MM-DD` 形式の文字列に変換する。
- `input()`：ユーザーから文字列を受け取る。
- `値A or 値B`：値Aが空文字列の場合、値Bを採用する。

## Git管理について

実行時の入力値はPython変数に格納されるため、Notebookのソースコードを変更する必要がない。

また、`nbstripout` を導入している場合、実行結果や実行回数による不要なGit差分を抑制できる。

```python
from datetime import date

today = date.today().isoformat()

start_date = input(f"開始日 [{today}]: ") or today
end_date = input(f"終了日 [{today}]: ") or today
```

```python
print(f"開始日: {start_date}")
print(f"終了日: {end_date}")
```
