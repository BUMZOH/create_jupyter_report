```python
import pandas as pd
from IPython.display import display


# サンプルデータ
df = pd.DataFrame(
    {
        "machine_no": ["MC001", "MC002", "MC003"],
        "production_qty": [1250, 980, 1420],
        "target_qty": [1200, 1200, 1300],
        "alarm_count": [3, 18, 25],
    }
)


# 表示用スタイル
styled_df = (
    df.rename(
        columns={
            "machine_no": "設備番号",
            "production_qty": "生産数",
            "target_qty": "目標数",
            "alarm_count": "アラーム件数",
        }
    )
    .style
    .format(
        {
            "生産数": "{:,}",
            "目標数": "{:,}",
            "アラーム件数": "{} 件",
        }
    )
    .hide(axis="index")
)

display(styled_df)
```
